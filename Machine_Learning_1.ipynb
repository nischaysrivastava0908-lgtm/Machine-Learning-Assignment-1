1. What is a parameter?

2. What is correlation? What does negative correlation mean?
Answer : Correlation is a statistical measure that describes the strength and direction of the relationship between two variables. It helps us understand how changes in one variable are associated with changes in another variable. Correlation ranges from -1 to +1. Positive correlation means both variables tend to increase together, while negative correlation means one tends to decrease as the other increases. Zero correlation indicates no linear relationship. However, correlation does not necessarily mean causation.

Negative correlation means that when one variable increases, the other variable tends to decrease, and vice versa. The correlation coefficient ranges from 0 to -1 for negative correlation. A value closer to -1 indicates a stronger negative linear relationship. For example, as the price of a product increases, its demand may decrease. However, negative correlation does not necessarily mean that one variable causes the other to change.

6. How do we handle categorical variables in Machine Learning? What are the common techniques?

Answer :-
Categorical variables in Machine Learning are handled by converting categories into numerical values so that algorithms can process them.

Common techniques:

    1. One-Hot Encoding: Converts each category into a separate binary column. Example: Red = [1, 0], Blue = [0, 1].

    2. Label Encoding: Assigns an integer to each category. Example: Cat = 0, Dog = 1. It is commonly used for target labels.

    3. Ordinal Encoding: Assigns numbers according to a meaningful order. Example: Low = 0, Medium = 1, High = 2.

    4. Binary Encoding: Converts categories into binary numbers. It can be useful when a feature has many unique categories.

    5. Target Encoding: Replaces each category with a statistic calculated from the target variable, using safeguards against data leakage.

16. What is an Optimizer? What are different types of optimizers? Explain each with an example.

Answer : An optimizer is an algorithm used in Machine Learning and Deep Learning to reduce the loss value by updating the model's parameters, such as weights and biases, during training. It helps the model learn from data and improve its predictions by finding suitable parameter values.

Different Types of Optimizers
1. Gradient Descent: Gradient Descent updates the model's parameters in the direction that reduces the loss function. It uses the entire training dataset to calculate the gradient for each update. For example, it adjusts the weights of a Linear Regression model to reduce prediction errors.

2. Stochastic Gradient Descent (SGD): SGD updates the model's parameters using one training example at a time. It can learn faster from large datasets, but its updates may fluctuate. For example, it updates the model weights after processing each image in an image-classification task.

3. Mini-Batch Gradient Descent: This optimizer updates the parameters using a small group of training examples called a mini-batch. It balances computational efficiency and stable learning. For example, a model may update its weights after processing 32 images at a time.

4. Momentum: Momentum improves Gradient Descent by using information from previous updates to help move consistently toward a minimum. It can reduce fluctuations and speed up training. For example, it helps a neural network move more smoothly toward lower loss values.

5. Adagrad: Adagrad adjusts the learning rate separately for each parameter. Parameters that receive frequent updates get smaller learning rates, while less frequently updated parameters can receive larger ones. For example, it can be useful in text-processing tasks where some words appear much less frequently than others.

6. RMSprop: RMSprop adjusts the learning rate using a moving average of recent squared gradients. This helps prevent the learning rate from becoming too small too quickly, as can happen with Adagrad. For example, it is often used to train neural networks, including recurrent neural networks.

7. Adam (Adaptive Moment Estimation): Adam combines the ideas of Momentum and RMSprop. It uses moving averages of gradients and squared gradients to adapt the learning rate for each parameter. For example, Adam is commonly used to train deep-learning models for image classification and natural language processing.

22. What is feature scaling? How does it help in Machine Learning?

Answer :
Feature scaling is a data preprocessing technique used in Machine Learning to bring numerical features into a similar range or scale. It prevents features with larger numerical values from dominating features with smaller values during model training.

Feature scaling helps Machine Learning algorithms learn more efficiently and can improve training speed and model performance. It is especially important for algorithms such as K-Nearest Neighbors (KNN), Support Vector Machines (SVM), and Gradient Descent-based models. For example, if a dataset contains age values ranging from 18 to 60 and salary values ranging from 20,000 to 100,000, feature scaling brings them to comparable scales.

Two common methods are Standardization, which scales values based on their mean and standard deviation, and Normalization, which often scales values to a range such as 0 to 1.
